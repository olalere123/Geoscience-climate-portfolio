```python
def capacity_clean(val):
    if pd.isna(val) or str(val).strip() == '':
        return np.nan
    val = str(val).strip()

    try:
        if '-' in val:
            min_val, max_val = val.split('-')
            min_val = float(min_val.strip())
            max_val = float(max_val.strip())
            return (min_val + max_val) /2
        else:
            return float(val)
    except:
            return np.nan

df2['clean_capacity']= df2['announced_capacity_(mt_co2/yr)'].apply(capacity_clean) 
df2.info()
```

```python
df2['sector'] = df['sector'].str.replace(" ", "_").str.replace("/", "_")
print(df2['sector'])
```

```python
df2['fate_of_carbon']=df2['fate_of_carbon'].str.replace(" ", "_").str.replace("/", "_").str.replace("S","s")
print(df2['fate_of_carbon'])
```

```python
project_count = df2.groupby(['sector', 'fate_of_carbon']).size().reset_index(name='project_count')
print(project_count.head(10))
```

```python
pivot = project_count.pivot(index='sector', columns="fate_of_carbon", values='project_count').fillna(0)
pivot.plot(kind='bar', figsize=(12, 6), colormap='Set2')
plt.title('Number of project by Sector and Fate of Carbon')
plt.xlabel('Sector')
plt.ylabel('Number of Project')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Fate of Carbon')
plt.tight_layout()
plt.savefig('Number of project by Sector and Fate of Carbon.png')
plt.show()
```

```python
pivot = project_count.pivot(index='sector', columns="fate_of_carbon", values='project_count').fillna(0)
plt.figure(figsize =(8,6))
sns.heatmap(pivot, annot = True, fmt=".0f", cmap ='YlOrRd', linewidth= 0.5)
plt.title('Heatmap of project by Sector and Fate of Carbon')
plt.xlabel('sector')
plt.ylabel('Fate Of Carbon')
plt.tight_layout()
plt.savefig('Heatmap of project by Sector and Fate of Carbon.png')
plt.show()
```
